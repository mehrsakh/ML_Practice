```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
plt.plot([1 , 2 , 3 , 4])
plt.show()
```

```python
plt.plot([1 , 3 , 4 , 9] , [-1 , 7 , 1 , 2] )
```

```python
plt.plot([1 , 5 , 8 , 7] , [4, 5 , 3 , -2] , 'ro')
```

```python
plt.plot([8 , 1 , 3 , 2] , [4 , 0 , -3 , 2] , 'b+')
```

```python
t = np.arange( 0 , 5 , 0.3)
plt.plot(t, t**2 )
```

```python
t = np.arange( 0 , 4 , 0.2)
plt.plot(t , t , 'r-')
```

```python
t = np.arange( 1 , 5 , 0.3)
plt.plot(t , t**2 , 'r--')
```

```python
t = np.arange( 0 , 5 , 0.4)
plt.plot(t , t , 'r-' , t, t**2 , 'bs' , t , t**3 , 'g*' )
```
